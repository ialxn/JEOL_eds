# JEOL Digi Line data (.pts)

EDX spectrum recorded for a single (horizontal) scan line). The data format and file extension are identical to elemental maps (`.pts`). `JEOL_eds.JEOL_DigiLine()` will complain if such data is provided.

Import required commands and define path were demo data is stored. Adapt this for your installation.

```python
from pathlib import PurePath
import numpy as np
from JEOL_eds import JEOL_DigiLine, JEOL_image
import JEOL_eds.utils as JU

base = PurePath("/home/alxneit/private/src/JEOL_eds")
```

Load example data (`data/DigiLine/View000_0000003.pts`).

```python
dl = JEOL_DigiLine(base / "data" / "DigiLine" / "View000_0000003.pts")
```

Again, next to the edx data meta data is stored in the file. The most important is the line that was (continuously) scanned.

```python
dl.scan_line
```

Typically, many scan lines are recorded and (individually) stored. In this case, the data contains 50 scans. Each line contains 256 point / spectra (scan area configured was 256x256) with each spectrum containing 4000 channels.

```python
dl.dcube.shape
```

Visualizing the position of the scan line is a bit more complicated. In the folder containing the scan line data an image file should also be present. This is the image on which the position of the scan line was defined. We load the image. This image is 512x512 pixels so we rebin the image by 2x2 to obtain the correct image dimensions (256x256). Finally, plot the line.

```python
im = JEOL_image(base / "data" / "DigiLine" / "View000_0000000.img")
rebinned = JU.rebin(im.image, (2,2))
JU.show_line(rebinned, (dl.scan_line, 0, dl.scan_line, dl.dcube.shape[1]))
```

For a quick overview, plot `JEOL_DigiLine.ref_spectrum`. This corresponds to the integral spectrum (all scan lines, all points).

```python
JU.plot_spectrum(dl.ref_spectrum, E_range=(0.0, 4.0), M_ticks=(4, None))
```

Plot the spectral map (energy versus position on scan line). Plot `np.sqrt()` of the intensity to enhance weak lines.

```python
m = dl.spectral_map(E_range=(0.0, 4.0), energy=True)
JU.plot_map(np.sqrt(m), color='red')
```

The interesting part is the feature around 3 keV (Pd). Extract and plot a line profile with x axis as nm. **Careful**, the calibration stored with the data uses the nominal magnification and is different from the one used by DM.

```python
x, I = dl.profile(interval=(2.8, 3.1), energy=True, xCalib=True)
JU.plot_profile(x, I, units='nm')
```
