# JEOL Point Line data (.pln)

Point Line data (series of points place at equal distance on a straight line. Do not confuse with `JEOL_eds.DigiLine()`).

Import required commands and define path were demo data is stored. Adapt this for your installation.

```python
from pathlib import PurePath
from JEOL_eds import JEOL_PointLine
import JEOL_eds.utils as JU

base = PurePath("/home/alxneit/private/src/JEOL_eds")
```

Load example data (`data/PointLine/View000_0000001.pln`).

```python
pl = JEOL_PointLine(base / "data" / "PointLine" / "View000_0000001.pln")
```

Again, next to the edx data meta data is stored in the file. The most important is the definition of the points (locations where spectra were collected). This data is stored in a `dict` and with marker (number) as key and a list `[FileName, xPos, yPos]` as content.

```python
pl.eds_dict
```

The image used to define the Point Line is stored as a `JEOL_eds.JEOL_image`.

```python
ref_img = pl.ref_image
```

The location of all measurement locations on the referece image can be plotted.

```python
pl.show_PointLine(color="yellow", ann_color="red")
```

Spectral data is available as `JEOL_PointLine.eds_data`. In this case 5 spectra with 4096 channels. The individual spectra (marker **n**) are plotted by `JU.plot_spectrum(pl.eds_data[n])`.

```python
JU.plot_spectrum(pl.eds_data[3], E_range=(0.0, 10.0), M_ticks=(4, None), log_y=True)
```
