# JEOL spectra (.eds)

These are spectra acquired by a spot scan or an area scan. Note, that the coordinates of the spot or the area selected is not saved in the `.eds` file (I have not figured out the location of this info). Easy solution: Take a photo of the definition which should suffice as, typically, you do not need the exact locations.

Import required commands and define path were demo data is stored. Adapt this for your installation.

```python
from pathlib import PurePath
from JEOL_eds import JEOL_spectrum
import JEOL_eds.utils as JU

base = PurePath("/home/alxneit/private/src/JEOL_eds")
```

Load example data (`data/spot.eds`).

```python
s = JEOL_spectrum(base / "data" / "spot.eds")
```

A `JEOL_spectrum` contains the spectral data (`JEOL_spectrum.data` and some meta data such as

- `JEOL_pts.file_date`: creation date and time (does change, when file is copied)
- `JEOL_pts.file_name`: file name
- `JEOL_pts.header`: additional meta data
- `JEOL_pts.footer`: additional meta data

```python
print(f"{s.file_name}: {s.file_name}")
```

Plot spectrum in the range 0 .. 10 keV. Add minor ticks (4 / 1) to x / y axis.

```python
JU.plot_spectrum(s.data, E_range=(0.0, 10.0), M_ticks=(4, 1))
```

You can export the sepctral data as a tab-separated text file. Use the parameter `E_range=` if you do not want to export the full range.

```python
JU.export_spectrum(s.data, base / "test.dat", E_range=(0.5, 10.0))
```
